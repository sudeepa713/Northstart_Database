```python
# Import required libraries for data cleaning and analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load all CSV files from the local data folder
customers = pd.read_csv("../data/customers.csv")
orders = pd.read_csv("../data/orders.csv")
deliveries = pd.read_csv("../data/deliveries.csv")
drivers = pd.read_csv("../data/drivers.csv")
vehicles = pd.read_csv("../data/vehicles.csv")
hubs = pd.read_csv("../data/hubs.csv")
complaints = pd.read_csv("../data/complaints.csv")
incidents = pd.read_csv("../data/incidents.csv")
app_events = pd.read_csv("../data/app_events.csv")
data_dictionary = pd.read_csv("../data/data_dictionary.csv")
```

```python
# Create backup copies so the original datasets remain unchanged
customers_clean = customers.copy()
orders_clean = orders.copy()
deliveries_clean = deliveries.copy()
drivers_clean = drivers.copy()
vehicles_clean = vehicles.copy()
hubs_clean = hubs.copy()
complaints_clean = complaints.copy()
incidents_clean = incidents.copy()
app_events_clean = app_events.copy()
```

```python
# Store all cleaned datasets in a dictionary for easier checking
datasets_clean = {
    "customers": customers_clean,
    "orders": orders_clean,
    "deliveries": deliveries_clean,
    "drivers": drivers_clean,
    "vehicles": vehicles_clean,
    "hubs": hubs_clean,
    "complaints": complaints_clean,
    "incidents": incidents_clean,
    "app_events": app_events_clean
}

# Display number of rows and columns in each dataset
for name, df in datasets_clean.items():
    print(name, df.shape)
```

```python
# Check missing values in each dataset
for name, df in datasets_clean.items():
    print("\n====================")
    print(name)
    print(df.isnull().sum())
```

```python
# Check duplicate records in each dataset
for name, df in datasets_clean.items():
    print(name, "duplicate rows:", df.duplicated().sum())
```

```python
# Standardise categorical text values by removing spaces and converting to lowercase

customers_clean["home_zone"] = customers_clean["home_zone"].str.strip().str.lower()
customers_clean["customer_type"] = customers_clean["customer_type"].str.strip().str.lower()
customers_clean["preferred_channel"] = customers_clean["preferred_channel"].str.strip().str.lower()
customers_clean["account_status"] = customers_clean["account_status"].str.strip().str.lower()

orders_clean["service_type"] = orders_clean["service_type"].str.strip().str.lower()
orders_clean["pickup_zone"] = orders_clean["pickup_zone"].str.strip().str.lower()
orders_clean["dropoff_zone"] = orders_clean["dropoff_zone"].str.strip().str.lower()
orders_clean["priority_level"] = orders_clean["priority_level"].str.strip().str.lower()
orders_clean["booking_channel"] = orders_clean["booking_channel"].str.strip().str.lower()

deliveries_clean["delivery_status"] = deliveries_clean["delivery_status"].str.strip().str.lower()

drivers_clean["base_zone"] = drivers_clean["base_zone"].str.strip().str.lower()
drivers_clean["employment_type"] = drivers_clean["employment_type"].str.strip().str.lower()
drivers_clean["shift_preference"] = drivers_clean["shift_preference"].str.strip().str.lower()

vehicles_clean["vehicle_type"] = vehicles_clean["vehicle_type"].str.strip().str.lower()
vehicles_clean["assigned_zone"] = vehicles_clean["assigned_zone"].str.strip().str.lower()
vehicles_clean["maintenance_status"] = vehicles_clean["maintenance_status"].str.strip().str.lower()
vehicles_clean["telematics_version"] = vehicles_clean["telematics_version"].str.strip().str.lower()

hubs_clean["zone"] = hubs_clean["zone"].str.strip().str.lower()
hubs_clean["hub_type"] = hubs_clean["hub_type"].str.strip().str.lower()

complaints_clean["complaint_type"] = complaints_clean["complaint_type"].str.strip().str.lower()
complaints_clean["channel"] = complaints_clean["channel"].str.strip().str.lower()
complaints_clean["severity"] = complaints_clean["severity"].str.strip().str.lower()
complaints_clean["status"] = complaints_clean["status"].str.strip().str.lower()

incidents_clean["incident_type"] = incidents_clean["incident_type"].str.strip().str.lower()
incidents_clean["severity"] = incidents_clean["severity"].str.strip().str.lower()
incidents_clean["resolution_status"] = incidents_clean["resolution_status"].str.strip().str.lower()

app_events_clean["event_type"] = app_events_clean["event_type"].str.strip().str.lower()
app_events_clean["device_type"] = app_events_clean["device_type"].str.strip().str.lower()
app_events_clean["zone_context"] = app_events_clean["zone_context"].str.strip().str.lower()
```

```python
# Convert date/time columns into proper datetime format

customers_clean["signup_date"] = pd.to_datetime(customers_clean["signup_date"], errors="coerce")

orders_clean["order_created_at"] = pd.to_datetime(orders_clean["order_created_at"], errors="coerce")

deliveries_clean["dispatch_time"] = pd.to_datetime(deliveries_clean["dispatch_time"], errors="coerce")
deliveries_clean["delivery_completed_at"] = pd.to_datetime(deliveries_clean["delivery_completed_at"], errors="coerce")

vehicles_clean["commission_date"] = pd.to_datetime(vehicles_clean["commission_date"], errors="coerce")

complaints_clean["created_at"] = pd.to_datetime(complaints_clean["created_at"], errors="coerce")

incidents_clean["reported_at"] = pd.to_datetime(incidents_clean["reported_at"], errors="coerce")

app_events_clean["event_timestamp"] = pd.to_datetime(app_events_clean["event_timestamp"], errors="coerce")
```

```python
# Fill missing numerical values using median values

customers_clean["loyalty_score"] = customers_clean["loyalty_score"].fillna(customers_clean["loyalty_score"].median())
customers_clean["app_engagement_score"] = customers_clean["app_engagement_score"].fillna(customers_clean["app_engagement_score"].median())

deliveries_clean["customer_rating_post_delivery"] = deliveries_clean["customer_rating_post_delivery"].fillna(
    deliveries_clean["customer_rating_post_delivery"].median()
)

complaints_clean["resolution_days"] = complaints_clean["resolution_days"].fillna(
    complaints_clean["resolution_days"].median()
)

complaints_clean["compensation_amount"] = complaints_clean["compensation_amount"].fillna(0)

app_events_clean["api_latency_ms"] = app_events_clean["api_latency_ms"].fillna(
    app_events_clean["api_latency_ms"].median()
)
```

```python
# Fill missing categorical/text values with "unknown"

customers_clean["preferred_channel"] = customers_clean["preferred_channel"].fillna("unknown")
orders_clean["priority_level"] = orders_clean["priority_level"].fillna("standard")
complaints_clean["status"] = complaints_clean["status"].fillna("unknown")
incidents_clean["resolution_status"] = incidents_clean["resolution_status"].fillna("unknown")
app_events_clean["device_type"] = app_events_clean["device_type"].fillna("unknown")
app_events_clean["zone_context"] = app_events_clean["zone_context"].fillna("unknown")
```

```python
# Calculate delivery duration in minutes
deliveries_clean["delivery_duration_minutes"] = (
    deliveries_clean["delivery_completed_at"] - deliveries_clean["dispatch_time"]
).dt.total_seconds() / 60

# Remove records where delivery duration is negative or missing
deliveries_clean = deliveries_clean[
    deliveries_clean["delivery_duration_minutes"].notna() &
    (deliveries_clean["delivery_duration_minutes"] > 0)
]
```

```python
# Create delay flag based on promised delivery window from orders

deliveries_clean = deliveries_clean.merge(
    orders_clean[["order_id", "promised_window_hours"]],
    on="order_id",
    how="left"
)

deliveries_clean["promised_window_minutes"] = deliveries_clean["promised_window_hours"] * 60

deliveries_clean["is_delayed"] = np.where(
    deliveries_clean["delivery_duration_minutes"] > deliveries_clean["promised_window_minutes"],
    1,
    0
)
```

```python
# Create binary flag for failed deliveries
deliveries_clean["is_failed"] = np.where(
    deliveries_clean["delivery_status"] == "failed",
    1,
    0
)
```

```python
# Create high-risk vehicle flag using actual column battery_health_pct
# Vehicle is considered high risk if battery health is below 70% or maintenance status is in repair

vehicles_clean["high_risk_vehicle"] = np.where(
    (vehicles_clean["battery_health_pct"] < 70) |
    (vehicles_clean["maintenance_status"] == "inrepair"),
    1,
    0
)
```

```python
# Convert complaint severity into numerical score

severity_map = {
    "low": 1,
    "medium": 2,
    "high": 3,
    "critical": 4
}

complaints_clean["severity_score"] = complaints_clean["severity"].map(severity_map)
```

```python
# Convert incident severity into numerical score

incidents_clean["incident_severity_score"] = incidents_clean["severity"].map(severity_map)
```

```python
# Create app failure flag from success_flag
# 1 means failed app event, 0 means successful app event

app_events_clean["app_event_failed"] = np.where(
    app_events_clean["success_flag"] == 0,
    1,
    0
)
```

```python
# Create high latency flag
# App event is considered high latency if API latency is greater than 300 milliseconds

app_events_clean["high_latency_event"] = np.where(
    app_events_clean["api_latency_ms"] > 300,
    1,
    0
)
```

```python
# Create integrated delivery dataset by joining deliveries with orders, customers, drivers, vehicles, and hubs

delivery_analysis_clean = (
    deliveries_clean
    .merge(orders_clean.drop(columns=["promised_window_hours"]), on="order_id", how="left")
    .merge(customers_clean, on="customer_id", how="left")
    .merge(drivers_clean, on="driver_id", how="left")
    .merge(vehicles_clean, on="vehicle_id", how="left")
    .merge(hubs_clean, on="hub_id", how="left")
)

delivery_analysis_clean.head()
```

```python
# Create integrated complaints dataset by joining complaints with customers and orders

complaints_analysis_clean = (
    complaints_clean
    .merge(customers_clean, on="customer_id", how="left")
    .merge(orders_clean, on="order_id", how="left")
)

complaints_analysis_clean.head()
```

```python
# Create integrated incidents dataset by joining incidents with deliveries, drivers, vehicles, and hubs

incidents_analysis_clean = (
    incidents_clean
    .merge(deliveries_clean, on="delivery_id", how="left")
    .merge(drivers_clean, on="driver_id", how="left")
    .merge(vehicles_clean, on="vehicle_id", how="left")
    .merge(hubs_clean, on="hub_id", how="left")
)

incidents_analysis_clean.head()
```

```python
# Check final dataset sizes after cleaning and merging

print("customers_clean:", customers_clean.shape)
print("orders_clean:", orders_clean.shape)
print("deliveries_clean:", deliveries_clean.shape)
print("drivers_clean:", drivers_clean.shape)
print("vehicles_clean:", vehicles_clean.shape)
print("hubs_clean:", hubs_clean.shape)
print("complaints_clean:", complaints_clean.shape)
print("incidents_clean:", incidents_clean.shape)
print("app_events_clean:", app_events_clean.shape)
print("delivery_analysis_clean:", delivery_analysis_clean.shape)
print("complaints_analysis_clean:", complaints_analysis_clean.shape)
print("incidents_analysis_clean:", incidents_analysis_clean.shape)
```

```python
# Validate important engineered columns

print(deliveries_clean[["delivery_duration_minutes", "is_delayed", "is_failed"]].head())
print(vehicles_clean[["battery_health_pct", "maintenance_status", "high_risk_vehicle"]].head())
print(complaints_clean[["severity", "severity_score"]].head())
print(app_events_clean[["api_latency_ms", "success_flag", "app_event_failed", "high_latency_event"]].head())
```
